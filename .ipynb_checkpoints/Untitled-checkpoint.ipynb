```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Define multi-index columns again
columns = pd.MultiIndex.from_tuples([
    ('Dataset', ''),
    ('Deploy #', ''),
    ('Non subset', 'Query-set'),
    ('Non subset', 'All-data'),
    ('Expert subset', ''),
    ('Unlabeled query set', 'Image-Align'),
    ('Unlabeled query set', 'Near-Nbors'),
    ('Unlabeled query set', 'CLIP-score'),
    ('Labeled query set', 'Match-Label'),
    ('Labeled query set', 'Match-Dist'),
])

# Data from GeoDE section only
data = [
    ['GeoDE (Acc)', 'Deploy 1', 0.872, 0.885, 0.921, 0.879, 0.880, 0.887, 0.882, 0.886],
    ['GeoDE (Acc)', 'Deploy 2', 0.450, 0.890, 0.910, 0.897, 0.892, 0.899, 0.900, 0.882],
    ['GeoDE (Acc)', 'Deploy 3', 0.950, 0.821, 0.850, 0.845, 0.760, 0.838, 0.830, 0.879],
    ['GeoDE (Acc)', 'Deploy 4', 0.827, 0.829, 0.845, 0.791, 0.783, 0.828, 0.841, 0.830],
]

# Create DataFrame
df = pd.DataFrame(data, columns=columns)

# Melt the dataframe to long-form
df_melted = df.melt(id_vars=[('Dataset', ''), ('Deploy #', '')],
                    var_name=['Category', 'Metric'],
                    value_name='Accuracy')

# Add a flag to mark expert subset
df_melted['Is Expert'] = df_melted['Category'] == 'Expert subset'

# Plot using seaborn
plt.figure(figsize=(12, 6))
sns.set(style="whitegrid")

# Use different styling for expert vs others
ax = sns.barplot(
    data=df_melted,
    x=('Deploy #', ''),
    y='Accuracy',
    hue='Category',
    palette={
        'Expert subset': 'tab:red',
        'Non subset': 'tab:blue',
        'Unlabeled query set': 'tab:green',
        'Labeled query set': 'tab:orange'
    },
    dodge=True,
    hue_order=['Expert subset', 'Non subset', 'Unlabeled query set', 'Labeled query set']
)

# Customize expert subset bars (e.g., hatch pattern)
for bar, is_expert in zip(ax.patches, df_melted['Is Expert']):
    if is_expert:
        bar.set_hatch('///')  # Add hatch
        bar.set_edgecolor('black')
        bar.set_linewidth(1.2)

# Final plot tweaks
plt.ylim(0.4, 1.0)
plt.title('GeoDE (Acc) - Accuracy by Strategy and Deployment')
plt.xlabel('Deployment')
plt.ylabel('Accuracy')
plt.legend(title='Strategy Category', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
